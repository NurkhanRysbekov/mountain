```python
import heapq
from collections import deque

карта = {
    "Бас_лагерь": [("Қарағайлы_орман", 4), ("Мұздық_асуы", 8), ("Тасты_жота", 3)],
    "Қарағайлы_орман": [("Мұздық_асуы", 2), ("Тау_шыңы", 7)],
    "Мұздық_асуы": [("Тау_шыңы", 3)],
    "Тасты_жота": [("Мұздық_асуы", 3), ("Тау_шыңы", 6)],
    "Тау_шыңы": []
}

h = {"Бас_лагерь": 8, "Қарағайлы_орман": 5, "Мұздық_асуы": 3, "Тасты_жота": 5, "Тау_шыңы": 0}

def dfs(start, goal):
    stack, seen, order = [(start, [start], 0)], set(), []
    while stack:
        curr, path, cost = stack.pop()
        if curr in seen: continue
        seen.add(curr); order.append(curr)
        if curr == goal: return path, cost, order
        for nxt, w in reversed(карта[curr]):
            if nxt not in seen: stack.append((nxt, path + [nxt], cost + w))

def bfs(start, goal):
    queue, seen, order = deque([(start, [start], 0)]), {start}, []
    while queue:
        curr, path, cost = queue.popleft()
        order.append(curr)
        if curr == goal: return path, cost, order
        for nxt, w in карта[curr]:
            if nxt not in seen:
                seen.add(nxt)
                queue.append((nxt, path + [nxt], cost + w))

def ucs(start, goal):
    pq, seen, order = [(0, start, [start])], set(), []
    while pq:
        cost, curr, path = heapq.heappop(pq)
        if curr in seen: continue
        seen.add(curr); order.append(curr)
        if curr == goal: return path, cost, order
        for nxt, w in карта[curr]:
            if nxt not in seen: heapq.heappush(pq, (cost + w, nxt, path + [nxt]))

def astar(start, goal):
    pq, seen, order = [(h[start], 0, start, [start])], set(), []
    while pq:
        _, cost, curr, path = heapq.heappop(pq)
        if curr in seen: continue
        seen.add(curr); order.append(curr)
        if curr == goal: return path, cost, order
        for nxt, w in карта[curr]:
            if nxt not in seen:
                g = cost + w
                heapq.heappush(pq, (g + h[nxt], g, nxt, path + [nxt]))

for name, func in [("DFS", dfs), ("BFS", bfs), ("UCS", ucs), ("A*", astar)]:
    path, cost, order = func("Бас_лагерь", "Тау_шыңы")
    print(f"{name}: Тексергені = {'->'.join(order)}      Жол = {'->'.join(path)}      Құны = {cost}")
```
